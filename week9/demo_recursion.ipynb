```python
# recursion: 1 function call itself
def print_numbers(n):
    # exit recursion if n < 0
    if n < 0:
        return
    # print number n
    print(n, end=' ')
    # call itself at smaller n
    print_numbers(n-1)
```

```python
print_numbers(10)
```

```python
def rec_sum(n):
    # base
    if n < 0:
        return 0
    # recursion
    return n + rec_sum(n-1)

def nor_sum(n):
    if n < 0:
        return 0
    s = 0
    for i in range(n+1):
        s += i
    return s
```

```python
print(rec_sum(10), nor_sum(10))
```

```python
# Viết hai hàm tính giai thừa n! theo đệ quy / for rồi thử lại
def nor_fact(n):
    if n == 0:
        return 1
    if n < 0:
        return None
    
    f = 1
    for i in range(2, n+1):
        f *= i
    
    return f

print(f'0! = {nor_fact(0)}, 1! = {nor_fact(1)}, 10! = {nor_fact(10)}')
```

```python
def rec_fact(n):
    if n == 0:
        return 1
    if n < 0: return None

    return n * rec_fact(n-1)

print(f'0! = {rec_fact(0)}, 1! = {rec_fact(1)}, 10! = {rec_fact(10)}')
```

```python
def rec_fibonacci(n):
    if n < 0:
        return None
    if n == 0 or n == 1:
        return 1
    
    return rec_fibonacci(n-1) + rec_fibonacci(n-2)
```

```python
for i in range(10):
    print(rec_fibonacci(i), end=' ')
```

```python
def nor_fibonacci(n):
    if n < 0:
        return None
    if n == 0 or n == 1:
        return 1
    
    count = 1
    f0, f1 = 1, 1
    while count < n:
        f2 = f0 + f1
        f0 = f1
        f1 = f2
        count += 1

    return f2
```

```python
for i in range(10):
    print(nor_fibonacci(i), end=' ')
```

```python
def fib_range(n):
    if n < 0:
        yield None

    f0, f1 = 1, 1
    count = 2

    if n == 0:
        yield f0
    else:
        yield f0
        yield f1
        while count <= n:
            f2 = f0 + f1
            yield f2
            f0 = f1
            f1 = f2
            count += 1
```

```python
for f in fib_range(10):
    print(f, end=' ')
```

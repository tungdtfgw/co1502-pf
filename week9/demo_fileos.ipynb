```python
import os
```

```python
print(os.curdir) # current directory
```

```python
print(os.pardir) # parent directory
```

```python
directories = os.listdir('.') # list contents in current dir
print(directories)
```

```python
parent_directories = os.listdir('..') # list contents in parent dir
print(parent_directories)
```

```python
for f in parent_directories:
    if os.path.isdir('../' + f):
        print(f'+ {f}')
    else:
        print(f'- {f}')
```

```python
def list_dir_v1(dir_name):
    # check if directory exist
    if not os.path.exists(dir_name):
        print(f'{dir_name} not exists.')
        return
    # check if dir_name is a directory
    if not os.path.isdir(dir_name):
        print(f'{dir_name} is not a directory.')
        return
    directories = os.listdir(dir_name)
    # print content of directory
    for f in directories:
        if os.path.isdir('../' + f):
            print(f'+ {f}')
        else:
            print(f'- {f}')
```

```python
dir_name = input('Enter a directory: ')
list_dir_v1(dir_name)
```

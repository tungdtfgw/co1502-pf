```python
try:
    a = int(input('Enter a: '))
    b = int(input('Enter b: '))
    c = a / b
    print(f'{a} / {b} = {c}')
except ZeroDivisionError:
    print('Cannot devide by zero!')
```

```python
a = int(input('Enter a: '))
b = int(input('Enter b: '))
if b != 0:
    c = a / b
    print(f'{a} / {b} = {c}')
else:
    print('Cannot devide by zero!')
```

```python
# open file 'demo.txt', mode is 'r' by default
f = open('demo.txt')
# read all content into a string
contents = f.read()
print(type(contents))
print(contents)
```

```python
# split contents to words
words = contents.split()
print(words[1], words[3], words[5])
```

```python
# split contents to lines
lines = contents.split('\n')
print(lines[1])
```

```python
# read to lines
f = open('demo.txt', 'r')
lines = f.readlines()
print(type(lines), len(lines))
print(lines[2])
```

```python
# read line by line (for large file)
f = open('demo.txt', 'r')
line = f.readline()
while line:
    print(line, end='')
    line = f.readline()
```

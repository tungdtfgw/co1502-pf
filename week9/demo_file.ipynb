```python
# open file 'demo.txt', mode is 'r' by default
f = open('demo.txt')
# read all content into a string
contents = f.read()
print(type(contents))
print(contents)
```

```python
# split contents to words
words = contents.split()
print(words[1], words[3], words[5])
```

```python
# split contents to lines
lines = contents.split('\n')
print(lines[1])
```

```python
# read to lines
f = open('demo.txt', 'r')
lines = f.readlines()
print(type(lines), len(lines))
print(lines[2])
```

```python
# read line by line (for large file)
f = open('demo.txt', 'r')
line = f.readline()
while line:
    print(line, end='')
    line = f.readline()
```

```python
f = open('demo2.txt')
contents = f.read()

def count_words(contents):
    words = contents.split(' ')
    return len(words)

def count_lines(contents):
    lines = contents.split('\n')
    return len(lines)

def count_specific_word(contents, word):
    words = contents.split(' ')
    count = 0
    for w in words:
        if w.lower() == word.lower():
            count += 1
    
    return count

from string import punctuation
def count_special_char(contents):
    count = 0
    for c in contents:
        if c in punctuation:
            count += 1
    
    return count
```

```python
print(f'Number of words: {count_words(contents)}')
print(f'Number of lines: {count_lines(contents)}')
print(f'Number of \'file\': {count_specific_word(contents, 'file')}')
print(f'Number of special chars: {count_special_char(contents)}')
```

```python
fruits = ['apple', 'banana', 'cherry', 'kiwi', 'mango', 'orange']
prices = [1.2, 0.5, 2.5, 1.0, 1.5, 0.8]

# Thêm một loại trái cây mới vào danh sách
def cheapest_fruit(fruits, prices):
    min_price = prices[0]
    i_min = 0
    # tìm vị trí mà giá nhỏ nhất
    for i in range(len(prices)):
        if prices[i] < min_price:
            min_price = prices[i]
            i_min = i
    # trả về loại trái cây tại vị trí đó
    return fruits[i_min]
```

```python
cheapest = cheapest_fruit(fruits, prices)
print(f'The cheapest fruit is: {cheapest}')
```

```python
# Ex2: Viết hàm nhập tên trái cây, trả về giá. Nếu không có trong danh sách, trả về None
def get_price(fruits, prices, fruit_name):
    for i in range(len(fruits)):
        if fruits[i] == fruit_name:
            return prices[i]
    
    return None # lệnh này không cần thiết, nhưng để rõ ràng hơn

fruit = input("Enter the name of the fruit: ")
price = get_price(fruits, prices, fruit)
if price is None:
    print(f"{fruit} is not in the list.")
else:
    print(f"The price of {fruit} is: {price}")
```

```python
# Ex3: Viết hàm in danh sách trái cây và giá tương ứng
def print_fruits(fruits, prices):
    print('+', '-' * 20, '+', '-' * 10, '+')
    print(f'| {"Fruit":<20} | {"Price":<10} |')
    print('+', '-' * 20, '+', '-' * 10, '+')

    for i in range(len(fruits)):
        print(f'| {fruits[i]:<20} | {prices[i]:>10} |')
    
    print('+', '-' * 20, '+', '-' * 10, '+')

print_fruits(fruits, prices)
```

```python
# Ex4: Viết hàm in các trái cây có giá lớn hơn trung bình
def over_price(fruits, prices):
    avg_price = sum(prices) / len(prices)
    
    over_fruits = []
    over_prices = []

    for i in range(len(prices)):
        if prices[i] > avg_price:
            over_fruits.append(fruits[i])
            over_prices.append(prices[i])

    print('Over average price fruits:')
    print(f'Average price: {avg_price:.2f}')
    print_fruits(over_fruits, over_prices)

over_price(fruits, prices)
```

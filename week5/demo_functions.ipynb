Function syntax

```python
def function_name(parameters):
    # function body
    return value # optional

````python
# Define a simple function: no parameters, no return value
def hello():
    print("Hello, world!")

# Call the function: mention the function name followed by parentheses
hello()
````

````python
hello()
````

````python
# A function with 1 parameter and no return value
def greet(name):
    print(f"Hello, {name}!")
````

````python
greet("Alice")
````

````python
name = input("Enter your name: ")
greet(name)
````

````python
greet() # error: missing 1 required positional argument: 'name'
````

````python
hello('Alice') # error: hello() takes 0 positional arguments but 1 was given
````

````python
def greet_v2(first_name, last_name):
    full_name = f"{first_name} {last_name}"
    print(f"Hello, {full_name}!")
````

````python
greet_v2("Alice", "Smith")
````

````python
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")
greet_v2(first_name, last_name)
````

````python
greet_v2('Alice')
````

````python
greet_v2('Alice', 'Smith', 'Extra') # error: greet_v2() takes 2 positional arguments but 3 were given
````

````python
# return value: a function can return a value to the outside using the return statement
def double(x):
    x = x * 2
    return x

a = 5
b = double(a)
print(f"a = {a}, b = {b}")
````

````python
# return can appear anywhere in the function, and it will immediately exit the function
# but a good practice is to have a single return statement at the end of the function
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

print(is_even(4))  # True
print(is_even(5))  # False
````

````python
n = int(input("Enter a number: "))
print(f'{n} is', 'even' if is_even(n) else 'odd')
````

````python
# Viết 1 hàm đổi nhiệt độ từ độ C sang độ F theo công thức: F = C * 9/5 + 32. Hàm nhận vào 1 tham số là nhiệt độ C và trả về nhiệt độ F.
````

````python
def convert_temperature(celsius):
    fahrenheit = celsius * 9/5 + 32
    return fahrenheit


c = float(input("Enter temperature in Celsius: "))
f = convert_temperature(c)
print(f"{c}°C is equal to {f}°F")
````

````python
# keyword arguments.
def hello_v2(name='World'):
    print(f"Hello, {name}!")

hello_v2()  # default argument ('World') is used
hello_v2("Alice")  # keyword argument is used
````

````python
def draw_triangle(n, char='*'):
    for i in range(n):
        for j in range(i+1):
            print(char, end=' ')
        print() # new line after each row
````

````python
draw_triangle(5)
````

````python
draw_triangle(4, '%')
````

````python
# Scope: local vs global variables.
n = 10 # global variable

def my_function(n):
    n = n + 5 # local variable
    print(f"Inside the function, n = {n}")

print(f"Before calling the function, n = {n}")
my_function(n)
print(f"After calling the function, n = {n}")
````

````python
# Sometimes you need a global variable inside a function.
n = 10 # global variable

def my_function():
    global n  # declare that we want to use the global variable n
    n = n + 5 # modify the global variable
    print(f"Inside the function, n = {n}")

print(f"Before calling the function, n = {n}")
my_function()
print(f"After calling the function, n = {n}")
````

````python
def swap(a, b):
    temp = a
    a = b
    b = temp
    print(f'a = {a}, b = {b}')

a = 5
b = 10
print(f'Before swap: a = {a}, b = {b}')
swap(a, b)
````

````python
print(f'After swap: a = {a}, b = {b}')
````

````python
def swap_v2(a, b):
    temp = a
    a = b
    b = temp
    return a, b

a = 5
b = 10
print(f'Before swap_v2: a = {a}, b = {b}')
a, b = swap_v2(a, b)
print(f'After swap_v2: a = {a}, b = {b}')
````

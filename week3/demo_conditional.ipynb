```python
a = int(input('Enter a: '))
b = int(input('Enter b: '))

op = input('Enter operation (+, -, *, /): ')

if op == '+':
    print(f'{a} + {b} = {a + b}')
elif op == '-':
    print(f'{a} - {b} = {a - b}')
elif op == '*':
    print(f'{a} * {b} = {a * b}')
elif op == '/':
    print(f'{a} / {b} = {a / b}')
else:
    print('Invalid operation')
```

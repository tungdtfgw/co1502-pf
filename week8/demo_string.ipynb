```python
# Ex: Ask user to enter username and password. Requirement:
# - username: contain only alphabet and numbers
# - password: contain aphabet, numbers and special character, upper case, lower case.

def valid_username(username):
    return username.isalnum()

def valid_password(password):
    # check upper
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False
    # check if password contains upper case
    for c in password:
        if c.isupper():
            has_upper = True
            break
    # check if password contains lower case
    for c in password:
        if c.islower():
            has_lower = True
            break
    # check if password contains a digit
    for c in password:
        if c.isdigit():
            has_digit = True
            break
    # check if password contains a special character
    for c in password:
        if not c.isalnum() and not c.isspace():
            has_special = True
            break
```

```python
username = input('Enter username: ')
password = input('Enter password: ')

if valid_username(username) and valid_password(password):
    print('Valid username and password')
else:
    print('Invalid username or password')
```

```python
def max_v1(a):         # avoid the built-in function max of Python
    max_item = a[0]
    for item in a:
        if item > max_item:
            max_item = item

    return max_item
```

```python
import random as rd
numbers = rd.sample(range(100), 10) # generate 10 random numbers 0-99
names = ['Binh', 'An', 'Lien', 'Duong', 'Linh', 'Chien']
print(numbers, names, sep='\n')
print(max_v1(numbers))
print(max_v1(names))
```

```python
def max_v2(a):         # avoid the built-in function max of Python
    max_item = a[0]
    max_i = 0
    n = len(a)
    for i in range(n):
        if a[i] > max_item:
            max_item = a[i]
            max_i = i

    return max_item, max_i
```

```python
n, i = max_v2(numbers)
print(f'Max = {n} at position {i}')
name, i = max_v2(names)
print(f'Max name = {name} at position {i}')
```

```python
def sum_v1(numbers):
    s = 0
    for n in numbers:
        s += n

    return s

print(sum_v1(numbers))
```

```python
def sum_v2(numbers):
    s = 0
    for i in range(len(numbers)):
        s += numbers[i]

    return s

print(sum_v2(numbers))
```

```python
def sum_evens(numbers):
    s = 0
    for n in numbers:
        if n %2 == 0:
            s += n

    return s

numbers = [1, 2, 3, 4, 5, 6]
print(sum_evens(numbers))
```

```python
def count_odds(numbers):
    count = 0
    for n in numbers:
        if n % 2 == 1:
            count += 1

    return count

numbers = rd.sample(range(100), 10)
print(numbers)
print(f'Number of odds: {count_odds(numbers)}')
```

```python
# TODO: Viết hàm đếm các số có chữ số tận cùng bằng 9
numbers = rd.sample(range(10, 100), 20)
print(numbers)

def count_unit9(numbers):
    count = 0
    for n in numbers:
        if n % 10 == 9:
            count += 1

    return count

print(f'Number of x9: {count_unit9(numbers)}')
```

```python
# TODO: Viết hàm đếm các tên quá dài (> 15 ký tự)
names = ['Ha Minh Lam', 'Le Dang Phong', 'Nguyen Tran Duc Minh', 'Luong Truong Giang', 'Do Minh An']
print(names)
print(len(names[0])) # độ dài của tên 'Ha Minh Lam'

def count_long_names(names, max_len):
    count = 0
    for name in names:
        if len(name) > max_len:
            count += 1

    return count

print(f'Number of long name: {count_long_names(names, 15)}')
```

```python
# check if x is in numbers
def is_found(numbers, x): 
    for n in numbers:
        if n == x:
            return True
        
    # only when for loop ends
    return False

print(numbers)
print(is_found(numbers, 36))
print(is_found(numbers, 360))
```

```python
# check where is x in numbers
def find_x(numbers, x):
    position = None # other programming language: position = -1
    for i in range(len(numbers)):
        if numbers[i] == x:
            position = i

    return position

print(numbers)
x = int(input('Enter x: '))
pos = find_x(numbers, x)

if pos is None:
    print(f'{x} Not found')
else:
    print(f'Found {x} at {pos}')
```

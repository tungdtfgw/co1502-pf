```python
# accessing list: index (starting from 0)
a = [1, 3, 5, 7, 9]
print(a[0]) # read
a[0] = 11 # write
print(a)
```

```python
# negative indexing: back from -1
print(a[-1])
```

```python
n = len(a) # return the length of a list
print(n)
print(a[-n]) # access the 1st item ~ a[0]
```

```python
# valid range of index: -n, -n + 1, ..., 0, 1, .., n-1
# try accessing an index ouside of this range => IndexError
print(a[n])
```

```python
# loop through items in a list
# using index
for i in range(n):
    print(f'a[{i}] = {a[i]}')
```

```python
# Simple algorithms with list: max, min, sum, count, find (exist, position), sort (asc / desc)
```

```python
def max_v1(a):         # avoid the built-in function max of Python
    max_item = a[0]
    for item in a:
        if item > max_item:
            max_item = item

    return max_item
```

```python
import random as rd
numbers = rd.sample(range(100), 10) # generate 10 random numbers 0-99
names = ['Binh', 'An', 'Lien', 'Duong', 'Linh', 'Chien']
print(numbers, names, sep='\n')
print(max_v1(numbers))
print(max_v1(names))
```

```python
def max_v2(a):         # avoid the built-in function max of Python
    max_item = a[0]
    max_i = 0
    n = len(a)
    for i in range(n):
        if a[i] > max_item:
            max_item = a[i]
            max_i = i

    return max_item, max_i
```

```python
n, i = max_v2(numbers)
print(f'Max = {n} at position {i}')
name, i = max_v2(names)
print(f'Max name = {name} at position {i}')
```

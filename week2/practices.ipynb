````python
# 1 cửa hàng cần thống kê doanh số của iphone 17 bán ra (new và like new).
# Giá của like new = 80% giá new.
# Nhập giá, số lượng máy new, số lượng máy like new, tính tổng doanh thu.
````

````python
n_news = int(input("Nhập số lượng máy new: "))
n_like_news = int(input("Nhập số lượng máy like new: "))
price = float(input("Nhập giá máy new: "))

total_revenue = n_news * price + n_like_news * price * 0.8
print(f"Tổng doanh thu: {total_revenue}")
````

````python
side = int(input('Enter iso triangle side length: '))
hypotenuse = side * (2 ** 0.5)
perimeter = side * 2 + hypotenuse
area = 0.5 * side ** 2
print(f'Perimeter: {perimeter:.2f}')
print(f'Area: {area:.2f}')
````

### Formated string
#### Alignment
- `<` : left align
- `>` : right align
- `^` : center align

For example:
```python
x = 123.456789
print(f"{x:<10.2f}") # left align, total 10 spaces, 2 decimal places
print(f"{x:>10.2f}") # right align, total 10 spaces, 2 decimal places
print(f"{x:^10.2f}") # center align, total 10 spaces, 2 decimal places
```

````python
x = 123.4567890123456789
print(f'|{x:<10.2f}|')
print(f'|{x:>10.2f}|')
print(f'|{x:^10.2f}|')
````

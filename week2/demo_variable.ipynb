### Variable
A variable is a representative of a value that can change.

A variable is referenced by a name => easy to remember and use.

Type of a variable is dependent on the value it holds.

```python
name = 'John'       # a variable of type string
distance = 100      # a variable of type integer
tax = 0.01          # a variable of type float
```

```python
# when a variable is referenced, the value it holds is returned
print(name)
print(distance)
print(tax)
```

### Naming rules for variable
- A variable name can contain letters, numbers, and underscores.
- A variable name must start with a letter or an underscore.
- A variable name cannot contain spaces or special characters.
- Avoid using reserved keywords as variable names (e.g., if, else, while, etc.).

Examples of valid variable names:
- name
- distance
- tax_rate
- a5
- _temp

Examples of invalid variable names:
- 1name (starts with a number)
- tax-rate (contains a special character)
- tax rate (contains a space)

### Naming conventions for variable
#### Should do
- Use lowercase letters for variable names.
- Use underscores to separate words in variable names (snake_case).

For example:
- name not Name
- distance not Distance
- tax_rate not TaxRate

#### Should not do
Do not use single letters for variable names, except for loop counters (e.g., i, j, k).
For example:
- Use distance instead of d
- Use tax_rate instead of t

#### Abbreviations
Some time, popular abbreviations can be used. For example:
- num for number
- table_no for table number
- VIP for very important person
- ID for identifier

#### Constants
Variables that are not meant to change can be defined as constants. By convention, constant names are written in uppercase letters with underscores separating words.
For example:
- PI = 3.14159
- TAX_RATE = 0.07

```python
print(type(name))      # <class 'str'>
print(type(distance))  # <class 'int'>
print(type(tax))       # <class 'float'>
```

```python
# Some other types of variables
numbers = [1, 2, 3]      # a variable of type list
is_valid = True         # a variable of type boolean
en_vn = {'one': 'một', 'two': 'hai'}  # a variable of type dictionary
print(type(numbers))  # <class 'list'>
print(type(is_valid))  # <class 'bool'> 
print(type(en_vn))  # <class 'dict'> 
```

```python
# type can be changed depending on the value assigned to the variable
a = 10
print(type(a))
a = 10.0
print(type(a))
a = '10'
print(type(a))
```

### Number operators:
- Addition: `+`
- Subtraction: `-`
- Multiplication: `*`
- Division: `/` (get float result)
- Floor division: `//` (get rounded integer result)
- Modulus: `%` (get remainder)
- Exponentiation: `**` (get power result)

```python
a = 10
b = 3
print(f'{a} + {b} = {a + b}')
print(f'{a} - {b} = {a - b}')
print(f'{a} * {b} = {a * b}')
print(f'{a} / {b} = {a / b}')
print(f'{a} // {b} = {a // b}')
print(f'{a} % {b} = {a % b}')
print(f'{a} ** {b} = {a ** b}')
```

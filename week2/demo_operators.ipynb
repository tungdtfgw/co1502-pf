### Number operators:
- Addition: `+`
- Subtraction: `-`
- Multiplication: `*`
- Division: `/` (get float result)
- Floor division: `//` (get rounded integer result)
- Modulus: `%` (get remainder)
- Exponentiation: `**` (get power result)

```python
a = 10
b = 3
print(f'{a} + {b} = {a + b}')
print(f'{a} - {b} = {a - b}')
print(f'{a} * {b} = {a * b}')
print(f'{a} / {b} = {a / b}')
print(f'{a} // {b} = {a // b}')
print(f'{a} % {b} = {a % b}')
print(f'{a} ** {b} = {a ** b}')
```

```python
# Input / Output
name = input("What is your name? ")
print(f'Hello {name}')
```

```python
a = input('Enter a: ') # a is a string not an integer
b = input('Enter b: ') # b is a string not an integer
print(a)
print(b)
```

```python
c = a + b # because a, b are strings, c is a concatenation of a and b, not the sum of a and b
print(c)
```

```python
a = int(input("Enter a: ")) # use int to convert string to integer
b = int(input("Enter b: "))
c = a + b
print(a, b, c, sep='\n')
```

### String operators:
- Concatenation: `+`
- Repetition: `*`
- Membership: `in` (check if a substring is in a string)

```python
s1 = 'Hello'
s2 = 'World'
print(s1 + ' ' + s2) # concatenation
```

```python
s = 'Hello'
print(s * 3) # repetition 3 times
print(3 * s) # repetition 3 times
```

```python
s1 = 'Hello World'
s2 = 'Hello'
s3 = 'hello' # h is lowercase

print(s2 in s1) # True
print(s3 in s1) # False
```

```python
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")
full_name = first_name + ' ' + last_name
print(f'Hello {full_name}!')
```

```python
n = int(input('Enter length of a bar: '))
c = input('Enter character to draw the bar: ')
bar = c * n
print(bar)
```

```python
# print a rectangle of width w and height h using character c
w = int(input('Enter width of the rectangle: '))
h = int(input('Enter height of the rectangle: '))
c = input('Enter character to draw the rectangle: ')
one_line = c * w + '\n'
rectangle = one_line * h
print(rectangle)
```

### Logical operators:
- `and`
- `or`
- `not`

```python
a = True
b = False
print(a and b) # False
print(a or b) # True
print(not a) # False
```

### Truth table for logical operators:

| a | b | a and b | a or b | not a |
|---|---|---------|--------|-------|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

```python
# Loop with range
for i in range(4):
    print(i, end=' ')
```

```python
for i in range(1, 5):
    print(i, end=' ')
```

```python
for i in range(10, 20):
    print(i, end=' ')
```

```python
for i in range(10, 20, 2): # range with step (start, stop, step)
    print(i, end=' ')
```

```python
for i in range(10, 0, -1): # reverse range (start, stop, -step)
    print(i, end='--')
```

```python
# print 'hello world' 5 times
for i in range(5):
    print('hello world')
```

```python
# print hello world 5 times with index
for i in range(1, 6):
    print(f'{i}: hello world')
```

```python
# if we don't need index, can use '_' as variable name
for _ in range(5):
    print('hello world')
```

```python
# Nested for loop
n = int(input("Enter n: "))
# print a square of n x n
for i in range(n):
    for j in range(n):
        print(n, end=' ')
    print() # new line after each row
```

```python
# Nested for loop
n = int(input("Enter n: "))
# print a square of n x n
for i in range(n):
    for j in range(n):
        print(j+1, end=' ')
    print() # new line after each row
```

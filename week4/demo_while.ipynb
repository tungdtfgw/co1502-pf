```python
# game odd or even
import random as rd
```

```python
playing = True
while playing:
    # generate a random number between 1 and 100 and ask the user to choose odd or even
    n = rd.randint(1, 100)
    answer = input('Is the number odd or even: ')
    # check if the answer is correct
    if answer == 'odd':
        if n % 2 == 1:
            print('You win!')
        else:
            print('You lose!')
    else:
        if n % 2 == 0:
            print('You win!')
        else:
            print('You lose!')
    # ask the user if they want to continue
    is_continue = input('Do you want to continue? (y/n): ')
    if is_continue == 'n':
        playing = False
```

```python
playing = True
while playing:
    n = rd.randint(1, 100)
    answer = input('Is the number odd or even: ')
    win = (answer == 'odd' and n % 2 == 1) or (answer == 'even' and n % 2 == 0)
    print('You win!' if win else 'You lose!')
    playing = input('Do you want to continue? (y/n): ') == 'y'
```
